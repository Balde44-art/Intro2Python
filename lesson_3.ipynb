
plt.subplot(2, 1, 1)

Cette commande sert à diviser la fenêtre d'affichage en une grille pour placer plusieurs graphiques côte à côte ou les uns sur les autres. Les trois chiffres indiquent la configuration :
• 2 : Le nombre total de lignes dans la grille.
• 1 : Le nombre total de colonnes dans la grille.
• 1 : L'index (la position) du graphique actuel. On cible ici la première case (le haut).

```python
#import math
#création de la fonction figure 
plt.figure()
#pour la creaion de plusieur graphiques 
plt.subplot(2,2,1)
plt.plot(x,y)
# la creation deux autre graphiques 

plt.subplot(2,2,2)
plt.plot(x,np.sin(x))
plt.plot(y,np.sin(y))
```

En réalité, plt (qui correspond au module matplotlib.pyplot) n'est pas une fonction, mais une interface procédurale (une collection de fonctions). Elle a été conçue pour imiter le fonctionnement de MATLAB.
Pour utiliser la véritable approche orientée objet de Matplotlib, on n'utilise plus directement plt.subplot() pour dessiner, mais on récupère et manipule des objets (des instances de classes) : l'objet Figure (la fenêtre globale) et l'objet Axes (la zone de dessin).

![Mon Schéma](ex9.png)

```python
fi,ax = plt.subplots(2,1,sharex  = True)
#ax.plot(x,y)
#plt.plot
ax[0].plot(x,y)
ax[1].plot(x,np.sin(x))
plt.show()
```

### visualiastion des foction mathématiques

```python
f = lambda x,y :np.sin(x)+np.cos(x+y)
```

La fonction numpy.meshgrid() sert à générer une grille de coordonnées à partir de plusieurs vecteurs unidimensionnels. Elle est principalement utilisée pour évaluer des fonctions mathématiques sur un espace à 2 ou 3 dimensions, ou pour tracer des graphiques en 3D (comme des surfaces ou des cartes de chaleur).

```python
X = np.linspace(0,5,100)
Y = np.linspace(0,5,100)
X,Y = np.meshgrid(X,Y)
Z = f(X,Y)
```

```python
#la présentation de la projection
ax = plt.axes(projection='3d')
ax.scatter(X,Y,Z,cmap='red')
```

Pour transformer une image en graphique (par exemple un graphique de l'intensité des pixels, un histogramme ou un tracé 3D) en Python, on utilise généralement les bibliothèques Matplotlib,

Pour représenter la façon dont une image numérique est structurée, on utilise généralement un repère à deux dimensions (X, Y) pour l'espace (les pixels) associé à un axe de valeurs (Z ou couleur) allant de 0 à 255 (qui représente l'intensité lumineuse ou la couleur en 8-bits).
Note : La valeur maximale standard en informatique est 255 (et non 225), ce qui correspond à \(2^8 = 256\) nuances de couleurs possibles (de 0 à 255).

```python
from scipy import misc 
face= misc.face(gray=True) 
#plt.imshow(face,cmap = 'gray')# il faut avoir une version de scipy <1.15.0
plt.hist(face.ravel(),bins = 255)
plt.show()
```
